```python
from langchain_community.document_loaders import WebBaseLoader
import bs4
loader2 = WebBaseLoader(web_path = 'https://onepiece.fandom.com/wiki/One_Piece',
                        bs_kwargs=dict(parse_only=bs4.SoupStrainer(
                            class_=("mw-page-title-main","mw-content-ltr mw-parser-output")
                        ))
                        )
text2 = loader2.load()
print(text2)
```

```python
from langchain_community.document_loaders import PyPDFLoader
loader3 = PyPDFLoader('attention.pdf')
text3 = loader3.load()
text3
```

```python
from langchain.text_splitter import RecursiveCharacterTextSplitter
text_splitter = RecursiveCharacterTextSplitter(chunk_size = 1000,chunk_overlap=200) # each chunk will overlap 200 characters with the previous one.
documents = text_splitter.split_documents(text3)
documents[:5]
```

```python
from langchain_ollama import OllamaEmbeddings
from langchain_community.vectorstores import Chroma
db = Chroma.from_documents(documents,OllamaEmbeddings(model = 'nomic-embed-text'))
```

```python
results = db.similarity_search("authors?")

results[0].page_content
```
